# Appels API

## Qu'est-ce qu'une API ?

API, ou Application Programming Interface, est un protocole de communication entre différentes applications web, ou différents composants de la même application. Ces composants différents partagent de l’information entre eux ou performer une action entre espaces.

En général, on les définit sur des points terminaux de l’application et elles doivent avoir un format d’échange d’information bien défini. Ces applications peuvent transporter les types d’objets suivants :

- XML
- JSON
- GraphQL
- Et d’autres types de données encapsulées…

Le but d’une API est la possibilité de **produire** ces structures en sortie afin de les envoyer et la possibilité de les **consommer** pour les traiter dans une application ou sous-application. La manière d’échanger ces données dépend du protocole utilisé.

Les APIs existent depuis 1991 et ont engendré plusieurs types de protocoles comme RPC et SOAP, mais le plus utilisé est le protocole REST, apparu en 2000.

# Les API REST

HTTP ou HyperText Transfer Protocol, est une convention que plusieurs API actuelles suivent. Il s’agit d’un protocole qui permet d’appeler à distance des fonctions d’un autre logiciel par le web, par le biais d'un client. Les API suivant cette convention sont des API **REST** (REpresentational State Transfer) et suivent ces règles :

- Identifier une ressource grâce aux URI (ex. base de données).
- Utiliser les **verbes HTTP** comme identifiants des opérations.
- Utiliser les **réponses HTTP** comme représentation des ressources.
- Les requêtes et réponses ont tous deux une en-tête et un corps.

Pour se fier à ce protocole, on définit une route sur l’API qui correspond à chaque verbe HTTP. Un verbe HTTP est une méthode de requête qui correspond au type d’action qui doit être performé :

- **GET :** obtenir une ressource
- **POST :** créer une nouvelle ressource
- **HEAD :** obtenir une ressource sans l’en-tête
- **PUT :** remplacer une ressource
- **PATCH :** mettre une ressource à jour
- **DELETE :** effacer une ressource

Chacune de ces actions renvoie une réponse HTTP. Ces réponses sont des codes de status variant de 100 à 500 et le premier chiffre représente la catégorie de réponse :

- **1xx :** Information
- **2xx :** succès
- **3xx :** redirection
- **4xx :** erreur côté client
- **5xx :** erreur côté serveur

Pour tester les fonctionnalités d’une API, on peut utiliser des applications clients telles que [Postman](https://www.postman.com/) ou [Insomnia](https://insomnia.rest/). [cURL](https://curl.se/) permet de faire la même chose en lignes de commande, et il existe des bibliothèques de fonctions pour chaque langage lorsqu'on veut mettre au point un consommateur d'API différent. Ces clients saisissent l'URL, la décomposent et utilisent chacune des parties pour obtenir la réponse. On peut considérer que le navigateur lui-même est un client API REST qui effectue des requêtes HTTP pour obtenir une page HTML en réponse.

- `http` et `https` signifient que le client utilise le protocole TLS vers le port 80 / 443.
- L'hôte, qui est le nom de domaine ou `localhost` en local, sera utilisé pour résoudre les adresses IP sur lesquelles se connecter.
- Toute route successive séparée par des `/` sera utilisée dans la requête pour dire au serveur ce qu'on souhaite obtenir (fichiers, pages HTML, réponse JSON d'API, etc...)

Côté Python, le moyen le plus simple d'effectuer des requêtes est d'utiliser la bibliothèques de fonctions [`requests`](https://requests.readthedocs.io/en/latest/), qui est basée sur la bibliothèque de fonctions `urllib3`.

Le principe est simple : `requests` utilise directement les verbes HTTP mentionnés ci-dessus et renvoie une réponse avec un code. Pour une simple requête **GET** :

````python
import json

import requests

r = requests.get("https://api.github.com/events")

print("Type d'objet obtenu: ", type(r))
print("Code de la réponse: ", r.status_code)

if r.ok:
    print("Réponse en texte brut:")
    print(r.text)
````

Le format obtenu provenant des routes d'API est très souvent du JSON, car facile à organiser et transporter. Avec les utilitaires `json` de Python, on peut observer le résultat, et même l'enregistrer dans un fichier.

````python
from pathlib import Path

if r.ok:
    api_example_path = Path("../data/api-example.json")
    api_example_path.parent.mkdir(parents=True, exist_ok=True)
    api_example_path.touch()
    api_example_path.write_text(json.dumps(r.json(), indent=2), encoding="utf-8")
    print("Listes de requêtes:")
    for element in r.json():
        print(element["id"])
````

## Sécurité

On peut aussi modérer l’API en limitant le nombre de requêtes par personne. En effet un trop grand nombre de requêtes surcharge ou ralentit le serveur. Pour limiter les abus, on génère une clé jeton par utilisateur. Les clés peuvent être associées à un compte utilisateur et distribuées de sorte à atteindre un quota de requêtes par heure. Une fois que ce quota est atteint, l’utilisateur peut redemander un jeton au bout d’un certain temps.

Il existe pour les API plusieurs protocoles permettant de protéger le service venant de clients peu scrupuleux. Les solutions actuelles incluent la création d’un serveur d’autorisation, qui fournit en général un jeton d’accès pour accéder à la donnée via l’API.

- **Authentification de base :** une méthode simple qui consiste à faire passer une paire nom d’utilisateur et mot de passe au sein de l’en-tête afin d’identifier l’utilisateur. Bien qu’elle soit explicite, elle n’est pas recommandée par mesure de sécurité.
- **Authentification par jeton :** le client fait passer un jeton dans l’en-tête de requête pour vérifier son identité et accéder aux ressources de l’API. L’avantage principal est que ce dernier n’a pas besoin d’être persisté en base de données.
- **JWT (JSON Web Token) :** une méthode puissante, populaire et sécurisée pour faire passer l’information entre deux camps dans un design d’API. Ces jetons représentent de l’information encodée qui peut être vérifiée à l’aide d’une signature digitale.
- **OAuth2 (OAuth 2.0) :** le serveur d’autorisation qui a standardisé la demande de jeton d’accès par le client. Il permet également l’accès au service via des applications tierces, comme la plupart des réseaux sociaux (ex. Google, Facebook, Github, etc…). Il définit 4 rôles : le propriétaire de la ressource (ex. données des employés, clients, livres, etc…), le client, le serveur de la ressource et le serveur d’autorisation.
- **OpenID Connect :** une extension d’OAuth2 qui essaie de le rendre plus interopérable.
- OpenAPI ou Swagger

La cellule suivante démontre un type de requête avec un mécanisme de clé API. L'API [OpenWeatherMap](https://openweathermap.org/) renseigne sur la météo selon une localisation. Les requêtes sont limitées par l'utilisation d'une clé API. Passer la mauvaise clé API renvoie un statut 401 (Non-autorisé).

Pour obtenir une clé API, il faut créer un compte sur https://openweathermap.org/, et aller dans la section **API keys** : https://home.openweathermap.org/api_keys.

Une clé API est une chaîne de caractères composée de chiffres et de lettres, générée d'après un nombre aléatoire généré et sécurisé cryptographiquement. Tout gestionnaire d'API doit permettre à l'utilisateur de
- Créer de nouvelles clés API
- Révoquer des clés API

Enfin, il est important de **ne jamais communiquer la clé API dans des projets de développement, de quelconque façon**.

````python
import requests

# Remplacer le contenu de cette ligne par la clé API
API_KEY = "xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx"

city = "Paris"

params = {
    "q": city,
    "appid": API_KEY,
    "units": "metric"
}

url = "http://api.openweathermap.org/data/2.5/weather"

response = requests.get(url, params=params)

if response.ok:
    data = response.json()
    print(f"Température à {city}: {data['main']['temp']}°C")
else:
    print("Code d'erreur :", response.status_code)
````

L'exemple ci-dessous décrit comment une requête API permet à l'utilisateur de créer un dépôt GitHub à distance, sans passer par l'interface graphique (soit le site, tout simplement). Cet exemple combine l'utilisation du verbe POST et l'identification par OAuth pour effectuer une telle action.

Comme pour l'API précédente, il est nécessaire d'avoir un compte Github en s'inscrivant sur la plateforme. Ensuite, il faut créer un nouveau jeton sur cette page : https://github.com/settings/personal-access-tokens 

- On peut lui donner un nom, une description et des permissions
- Pour cette tâche, le jeton doit avoir accès à tous les dépôts (cocher All Repositories) et doit avoir les permissions "Administration" et "Repository Creation", les deux en écriture

Une fois le jeton généré, le copier et le conserver précieusement (ex. fichier texte).

L'autorisation OAuth fonctionne en modifiant les en-têtes de la requête. Un en-tête (header) permet d'identifier l'émetteur de la requête à l'aide de champs qui sont décrits par la documentation API. Dans le cas de l'autorisation OAuth, l'en-tête doit contenir le jeton API.

Le verbe POST quant à lui accepte une route, et un contenu à faire passer dans le corps. La nature de ce corps peut différer en fonction de l'API (ex. résultats de formulaire), mais ici, il s'agit principalement de JSON.

*Note:* `requests` utilise d'habitude `data` dans sa fonction `post`. `json` ici est simplement un raccourci pour éviter d'épeler `data=json.dumps(body)`.

````python
GITHUB_TOKEN = "github_pat_xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx"
headers = {
    "Accept": "application/vnd.github+json",
    "Authorization": f"Bearer {GITHUB_TOKEN}",
    "X-Github-Api-Version": "2026-03-10",
}

url = "https://api.github.com/user/repos"
body = {
    "name": "api-generated-repo",
    "description": "This repo has been generated from the API!",
    "private": True,
    "is_template": False,
    "homepage": "https://github.com",
}

r = requests.post(url, json=body, headers=headers)

if r.ok:
    print("Repo créé avec succès !")
    print(f"URL: {r.json()['html_url']}")
else:
    print(f"Erreur: {r.status_code} - {r.text}")
````
